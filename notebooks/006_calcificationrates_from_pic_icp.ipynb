```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from pace_pic_validation.io.calcification_pic_processing import pic_from_ca
```

```python
pic_icp_path = '/Users/timfugl/Desktop/PhD_Ocean_Color_Chemistry/Obj2_CalcificationRates_of_Coccoblooms/Data/SS20260601_ICP_OES_isolated.csv'

ds = pd.read_csv(pic_icp_path)

ds_t0_sur = ds[ds['Sample ID'].str.endswith('sur_t0')]

ds = ds[~ds['Sample ID'].str.endswith('sur_t0')]

print(len(ds_t0_sur))

# Simply concatenate the DataFrame with itself
ds_t0_sur = pd.concat([ds_t0_sur, ds_t0_sur], ignore_index=False)

ds_A = ds_t0_sur[0:12]
#print(ds_A)
ds_B = ds_t0_sur[12:24]

ds_A['Sample ID'] = ds_A['Sample ID'].str.replace('sur_t0', 'A_sur_t0')
ds_B['Sample ID'] = ds_B['Sample ID'].str.replace('sur_t0', 'B_sur_t0')

ds_t0_sur = pd.concat([ds_A, ds_B], ignore_index=False)

ds = pd.concat([ds, ds_t0_sur], ignore_index=False)


ds['Sample ID'] = ds['Sample ID'].str.replace(' ', '')


ds['Sample ID'] = ds['Sample ID'].str.replace('_t1', '')
ds['Sample ID'] = ds['Sample ID'].str.replace('_t0', '')

ds_t0 = ds[ds['time ID'] == 't0']
ds_t1 = ds[ds['time ID'] == 't1']

print(ds_t0)





print(ds_t0.columns)


print(ds_t1)
print(ds_t0)
```

```python
ds_t0 = pic_from_ca(ds_t0, 'Ca mg_L', 'Na mg_L', 'V_acid', 'V_filt_mL', 'PIC [mmol/m3]')
ds_t1 = pic_from_ca(ds_t1, 'Ca mg_L', 'Na mg_L', 'V_acid', 'V_filt_mL', 'PIC [mmol/m3]')



merged_df = pd.merge(ds_t0, ds_t1, on='Sample ID', suffixes=('_t0', '_t1'))

merged_df['Delta PIC [mmol/m3]'] = merged_df['PIC [mmol/m3]_t1'] - merged_df['PIC [mmol/m3]_t0']

#merged_df = merged_df[~(merged_df['Sample ID'].isin(['SS20260601_006_C_dep', 'SS20260601_006_D_dep']))]









```

```python
merged_df.plot(kind='bar', x='Sample ID', y='Delta PIC [mmol/m3/day]')

plt.show()
```
