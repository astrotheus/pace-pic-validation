```python
import pandas as pd

from pace_pic_validation.io.calcification_pic_processing import pic_from_ca
```

```python
sample_sheet = '/Users/timfugl/Desktop/PhD_Ocean_Color_Chemistry/Research_Campaigns/Norbjørn /August 2026/data/NB20260814_sample_data.csv'
```

```python
df = pd.read_csv(sample_sheet)

print(df.columns)
```

```python
df_new = pic_from_ca(df, ca_column = 'Ca (mg/L)', na_column = 'Na (mg/L)', V_acid_column = 'V_acid_PIC (mL)', V_filt_column = 'V_filtered_PIC (mL)', end_col_name = 'PIC_in_mmol_m3')
```

```python
print(df_new.columns)
```

```python
mask = (df_new['lon_end'] == 0) | (df_new['lat_end'] == 0)

mask = (
    df_new['lat_end'].isna()
    | df_new['lon_end'].isna()
    | (df_new['lat_end'] == 0)
    | (df_new['lon_end'] == 0)
)

df_new['lat'] = df_new['lat_start'].where(
    mask,
    (df_new['lat_start'] + df_new['lat_end']) / 2
)

df_new['lon'] = df_new['lon_start'].where(
    mask,
    (df_new['lon_start'] + df_new['lon_end']) / 2
)
```

```python
df_new = df_new.drop(columns = ['V_filtered_PIC (mL)', 'V_filtered_SEM (mL)', 'V_filtered_POC (mL)',
       'V_filtered_TPC (mL', 'V_acid_PIC (mL)', 'Ca (mg/L)', 'Na (mg/L)', 'lat_start', 'lon_start', 'lat_end', 'lon_end'])


df_new = df_new[['Sample ID', 'Cruise', 'Station', 'datetime (UTC)','lat', 'lon', 'turb_1', 'turb_2',
       'turb_3','PIC_in_mmol_m3', 'Chl A (from Ferrybox Flourescence, ug/L)', 'observations']]



df_new.to_csv('/Users/timfugl/Desktop/PhD_Ocean_Color_Chemistry/Research_Campaigns/Norbjørn /August 2026/data/NB20260814_PIC_discrete.csv', index = False)
```
